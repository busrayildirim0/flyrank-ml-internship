# 2. The question: decision, action, and cost

## Research question

**Which pages should be reviewed first for a potential content refresh, expansion, protection, pruning, or monitoring action, based on observable search and content signals?**

### Decision

The decision is which pages should receive limited content-review capacity first.

The system should rank pages by their evidence of potential opportunity or risk rather than simply labeling a page as "good" or "bad."

### Unit of analysis

The primary unit of analysis will be **one content page** (`content_hash_id`).

For the warehouse work, I will aggregate daily observations into page-level features over explicitly defined time windows.

### Output

The output will be a **ranked review queue**. Each candidate should include:

- a priority score or ranking;
- a suggested action category;
- reason codes explaining why the page was ranked highly;
- a confidence or evidence indicator.

### Action

A human content or SEO reviewer could use the ranking to decide which pages to inspect first and whether to consider actions such as:

- refresh existing content;
- expand missing or weak sections;
- protect a strong page;
- investigate possible consolidation;
- prune a low-value page;
- monitor the page rather than immediately changing it.

### Cost of a wrong recommendation

A false positive can waste editorial or SEO effort on a page that did not need intervention.

A false negative can be more costly when an important page is genuinely losing visibility and the opportunity to review it is missed.

Therefore, the project should prioritize a useful ranking for limited review capacity rather than maximizing generic classification accuracy.

The system will support human review; it will not claim that a page is guaranteed to recover after a refresh.

```python
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

print(f"Rows: {len(df):,}")
print(f"Columns: {df.shape[1]}")

declining_rate = (
    df["trend_direction"].str.lower().eq("down").mean()
)

print(f"Declining rate: {declining_rate:.3f}")

print(
    "Search volume / impressions correlation:",
    round(df["search_volume"].corr(df["impressions_90d"]), 3)
)
```

### What these numbers suggest

The starter dataset contains 30,000 content pages.

Under the starter definition, approximately **54.2%** of pages are labeled as declining. This makes prioritization potentially useful because a reviewer cannot realistically investigate every page manually.

The correlation between `search_volume` and `impressions_90d` is approximately **0.001**, which is effectively zero in this starter sample. This suggests that keyword search volume alone is not sufficient to explain how much search exposure an individual page receives.

Earlier in the starter experiments, the random forest reached approximately **0.740 Precision@50**, compared with **0.240** for the hand-written baseline. This provides an initial indication that combining multiple observable signals may be more useful than relying on a single simple rule.

These observations motivate further investigation with the larger warehouse dataset, but they do not prove that any individual signal causes traffic changes or that a refresh will produce recovery.
